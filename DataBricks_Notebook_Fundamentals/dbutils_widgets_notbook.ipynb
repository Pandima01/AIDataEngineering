```python
print("can you get the value of the widget using dbutils.widgets.get and store into a local python variable tblname")
tblname=dbutils.widgets.get("tablename")
print("user passed the value of ?",tblname)
```

```python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.dropdown("dropdown_widget","PANDIMA",["PANDIMA","PRAVEEN","YOSNITH","CHANGAMITHIRAN"],"Select your name")
aspirant_name_chosen=dbutils.widgets.get("dropdown_widget")
print("Good morning",aspirant_name_chosen)
```

```python
dbutils.widgets.multiselect("multiselect_widget","wd36",["wd32","we43","we45","wd36","wd38","we49"],"Select your team name")
all_batches=dbutils.widgets.get("multiselect_widget")
print(all_batches)
print(type(all_batches))
all_batches_list=all_batches.split(",")
print(all_batches_list)
print(type(all_batches_list))
batch1=all_batches_list[0]
print(batch1)
batch2=all_batches_list[1]
print(batch2)
batch3=all_batches_list[2]
print(batch3)
```
