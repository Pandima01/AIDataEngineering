How to run a DBFS (like Hadoop) FS commands inside a notebook using %fs magic command to copy the uploaded data into some other volume from the uploaded volume**

- this %fs ls is equalent to **hadoop fs -ls ** command we used in hadoop
- this %fs ls is listing DBFS files/folders which is a wrapper on top of some Cloud Storage (Cloud Datalake)

```python
%fs ls "dbfs:///Volumes/workspace/default/volumewd38_datalake/dbfs_cloudfs_data/"
```
