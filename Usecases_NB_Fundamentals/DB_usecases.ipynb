```python
%python

# Read CSV file from the volume
df1 = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv")
)

# Display the dataframe
display(df1)
```

```python
%python

from pyspark.sql.functions import col

# Rename columns
df1_clean = (
    df1
    .withColumnRenamed("Mobile Operating System", "mobile_operating_system")
    .withColumnRenamed("Percent of Usage", "percent_of_usage")
)

# Verify column names
print(df1_clean.columns)

# Save as Databricks table
df1_clean.write \
    .mode("overwrite") \
    .saveAsTable("default.mobile_os_usage")

print("Table created successfully!")

```
